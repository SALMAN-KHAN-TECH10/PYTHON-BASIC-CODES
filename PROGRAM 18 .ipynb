### PYTHON PROGRAM TO FIND THE SUM OF ALL NATURAL NUMBER 

```python
num = int(input("enter the number "))
if num < 0 :
    print("please enter a positive number")
else:
    sum = 0
    while num > 0:
        sum += num
        num -= 1
    print(f"the sum of the number is {sum}")
        
```
