PROGRAM TO ADD TWO NUMBERS 

```python
# SOLUTION WITH PRE DEFINE VARIABLE

NUM1 = 29
NUM2 = 34
SUM = NUM1+NUM2
print(f" THE SUM OF  {NUM1} AND {NUM2} IS = {SUM } ")
```

```python
# SOLUTION WITH USER INPUT

NUM1 = int(input("ENTER THE FIRST NUMBER"))
NUM2 = int(input("ENTER THE SECOND NUMBER "))
SUM = NUM1+NUM2
print(F"THE SUM OF {NUM1} AND {NUM2} IS = {SUM}")
```
