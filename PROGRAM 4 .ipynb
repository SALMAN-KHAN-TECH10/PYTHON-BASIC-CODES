PYTHON PROGRAM TO SWAP TWO NUMBER 

```python
#  SOLUTION 1 USING THE THIRD VARIABLE 
NUM1 = 18
NUM2 = 34
TEMP = NUM1
NUM1 = NUM2
NUM2 = TEMP
print(f"after swapping the values of num1 IS {NUM1} and num2 are: {NUM2}")
```

```python
# SECOND SOLUTION WITHOUT USING THE THIRD VARIABLE
NUM1 = 12
NUM2 = 13
NUM1,NUM2 = NUM2 , NUM1
print(f"the values after swapping the values of num1 is {NUM1} and num2 is {NUM2}")
```
