### PROGRAM TO FIND HCF/GCD OF TWO NUMBERS 

```python
def find_HCF(x, y):
    if x > y:
        smaller = y
    else:
        smaller = x
    for i in range(1, smaller + 1):
        if (x % i == 0) and (y % i == 0):
            HCF = i
    return HCF

print(f"The HCF of two numbers is :: {find_HCF(12, 30)}")
```
