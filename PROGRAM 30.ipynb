##### PYTHON PROGRAM TO FIND THE FACTORIAL OF THE NUMBER THROUGH RECURSION 

```python
num=int(input("eter the number "))
fact=1
def fact(num):
    if num == 1:
        return num
    else:
        return (num*fact(num-1))
if num <=0:
    print("please enter a positive number ")
else:
    print(f"the factorial of the given {num} number is {fact(num)}")
```
