### Python Program to Illustrate Different Set Operations (Hindi)

```python
set1 ={1,2,3,4,5,6,7,8}
set2 ={5,6,7,8,9,10,11,12}

print(f"the union of the set1 and set2 is :: {set1 | set2}")

print(f"the intersction of the set1 and set2 is :: {set1 & set2}")

print(f"the difference of the set1 and set2 is :: {set1 - set2}")

print(f"the symetric difference of the set1 and set2 is :: {set1 ^ set2}")
```
