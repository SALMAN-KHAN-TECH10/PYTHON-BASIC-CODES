#### PYTHON PROGRAM TO CONVERT DECIMAL INTO BINARY USING RECURSION 


```python
num = int(input("enter the number here "))
def convert_binary(num):
    if num > 1:
        convert_binary(num//2)
    print(num%2 , end=" ")
print(f"the binary of the given number {num} is :: ")
convert_binary(num)
```
