### HOW TO TRANSPOSE A MATRICES IN PYTHON PROGRAMMING 

```python
# SOLUTION NO 1 USING FOR LOOP 
A =[[2,5,3],
    [5,3,7]]

T =[[0,0],
    [0,0],
    [0,0]]

for i in range(len(A)):
    for j in range(len(A[0])):
        T [j][i] = A [i][j]

for i in T:
    print(i)


```
