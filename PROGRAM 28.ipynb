### PROGRAM TO DISPLAY FIBONACCI SEQUENCE USING RECURSION 

```python
n = int(input("enter a positive number :: "))
def fibonacci(n):
    if n <= 1:
        return n 
    else:
        return(fibonacci(n-1)+fibonacci(n-2))

if n<=0:
    print("please enter a positive number ")
else:
    print("fibonacci sequence ::")
    for i in range(n):
        print(fibonacci(i))
```
