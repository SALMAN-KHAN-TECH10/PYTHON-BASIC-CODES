#### PYTHON PROGRAM TO CONVERT DECIMAL TO BINARY , OCTAL AND HEXADECIMAL


```python
num = int(input("enter the number "))
print(f"the conversion of decimal a number {num} is :: ")
print(bin(num) ,"for binary conversion")
print(oct(num), "for octal conversion")
print(hex(num) , "for hexadecimal conversion")
```
