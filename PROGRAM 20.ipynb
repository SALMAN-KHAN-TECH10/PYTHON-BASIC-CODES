### FIND THE NUMBER DIVISIBLE BY ANOTHER NUMBER 

```python
print("the number which is divisible by  13  are ::")
for i in range(1,100):
    if i % 18 == 0:
        print(f"the number which is divisible by 13 are :: {i}")
```

```python
#  SECOND METHOD USING LAMBDA FUNCTION
lst=[13,26,78,55,37,88,90,91,67]
result=list(filter(lambda x : x % 13 == 0 ,(lst)))
print(f"the numbers which are divisible by 13 are :: {result}")
```
