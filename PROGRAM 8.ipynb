PROGRAM TO CHECK A LEAP YEAR 

```python
YEAR=int(input("ENTER THE YEAR "))
if (YEAR % 400 ==0 ) and (YEAR % 100 == 0):
    print(f"{YEAR} IS A LEAP YEAR ")
elif(YEAR % 4 ==0) and (YEAR % 100 !=0):
    print(f"{YEAR} IS A LEAP YEAR ")
else:
    print(f"{YEAR} IS NOT A LEAP YEAR ")
    
```
