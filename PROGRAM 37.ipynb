## How to Remove Punctuations From a String

```python
punc = '''!@#$%&*()_-+={[}]|\;:"'<,>.?/~`'''
string = input("enter anything here ::")
empty_string = ""
for i in string :
    if i not in punc :
        empty_string = empty_string + i
print(f"the text enter by the user is :: {string} \n is :: {empty_string}")
```
