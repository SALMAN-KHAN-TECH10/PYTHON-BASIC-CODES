TO CONVERT KILOMETER INTO MILES 

```python
# SOLUTION 1 KILOMETER TO MILES 

KILOMETER = float(input("ENTER THE VALUE IN KILOMETER "))
MILES = KILOMETER * 0.621371
print(f"{KILOMETER} KILOMETER WILL BE  {MILES} MILES ")
```
