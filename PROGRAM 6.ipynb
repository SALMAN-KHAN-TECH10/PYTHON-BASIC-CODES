TO CHECK WHETHER NUMBER IS POSITIVE NEGATIVE OR 0 

```python
num = int(input("enter the number "))
if num > 0:
    print("NUMBER IS POSITIVE ")
elif num < 0:
    print("THE NUMBER IS NEGATIVE ")
elif num == 0:
    print("THE NUMBER IS ZERO ")
else:
    print("INVALID INPUT ")
```
