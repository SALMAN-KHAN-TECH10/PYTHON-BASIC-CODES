## Program to Count the Number of Each Vowel

```python
text = " Harry Potter and the the Prince of Persia "
vowels = "aeiou"
text = text.casefold()
print(text)
count = {}.fromkeys(vowels,0)

for char in text:
    if char in count:
        count[char]+=1
print(f"the number of each vowels in the text is {count}")
```

```python
# SECOND METHOD USING LIST AND DICT COMPREHENSION
text = " Harry Potter and the the Prince of Persia "
vowels = "aeiou"
text = text.casefold()
print(text)

count = {key:sum([1 for char in text if char == key ]) for key in vowels }
print(count)
```
