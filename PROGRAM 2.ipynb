PYTHON PROGRAM TO FIND THE SQUARE ROOT

```python
# FIRST SOLUTION USING EXPONENTIATION 

NUM1 = int(input("ENTER THE  NUMBER"))
SQUARE_ROOT = NUM1 ** (0.5)
print("THE SQUARE ROOT OF THE GIVEN NUMBER IS ",SQUARE_ROOT)
```

```python
# SECOND SOLUTION USING MATH MODULE 

import math
NUM1 = int(input("ENTER THE  NUMBER"))
SQUARE_ROOT = math.sqrt(NUM1)
print("THE SQUARE ROOT OF THE GIVEN NUMBER IS ", SQUARE_ROOT)
```
