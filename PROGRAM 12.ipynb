PROGRAM TO PRINT ALL PRIME NUMBER IN AN INTERVAL 

```python
lower = int(input("enter the number "))
upper = int(input("enter the number "))
for num in range (lower , upper+1):
    if num  > 1:
        for i in range (2,num):
            if num % i == 0:
                break
        else:
                print(num)
                
```
