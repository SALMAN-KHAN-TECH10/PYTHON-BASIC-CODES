### FIBONACCI SEQUENCE IN PYTHON PROGRAMMING 

```python
a= 0
b = 1
num=int(input("enter the number of term "))
if num <= 0:
    print(a)
else:
    print(a)
    print(b)
for i in range (2,num+1):
        c = a + b
        a = b 
        b= c
        print(c)
```
