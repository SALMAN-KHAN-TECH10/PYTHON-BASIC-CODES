### MAKE A SIMPLE CALCULATOR IN PYTHON PROGRAMMING 

```python
num1= int(input("enter the first number ::"))
num2 =int(input("enter the second number ::"))
print("press 1 for addation \n press 2 for subtraction \n press 3 for multiplication \n press 4 for division ")
choice = int(input("enter your choice option ::"))
if choice == 1 :
    print(f"the addation of the given  number {num1} and {num2} is :: {num1 + num2 }")
elif choice == 2:
    print(f"the subtraction of the given  number {num1} and {num2} is :: {num1 - num2}")
elif choice == 3:
    print(f"the multiplication of the given  number {num1} and {num2} is :: {num1 * num2 }")
elif choice == 4 :
    print(f"the division of the given  numbwr {num1} and {num2} is :: { num1 / num2}")
else:
    print("please enter the valid option ")
```
