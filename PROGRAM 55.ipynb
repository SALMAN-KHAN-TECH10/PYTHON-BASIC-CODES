## How to Convert String to DateTime using Python Codes

```python
# SOLUTION 1 USING DATE TIME MODULE 

from datetime import datetime
date = " dec 3 2003  9:30 "
date_time = datetime.strptime(date, " %b %d %Y %I:%M ")
print(date_time)
```

```python
# SOLUTION 2 USING DATEUTIL MODULE 

from dateutil import parser
date_time = parser.parse(" dec 4 2003  9:30 ")
print(date_time)
```
