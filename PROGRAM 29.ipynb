### SUM OF ALL NATURAL NUMBER USING RECURSION IN PYTHON 



```python
num = int(input("enter the number "))
def nnn(num):
    if num <= 1:
        return num 
    else:
        return (num+nnn(num-1))
if num <=0:
    print(("please enter an positive number "))
else:
    print(f"the sum of the all natural numbers till {num} is {nnn(num)}")
```
