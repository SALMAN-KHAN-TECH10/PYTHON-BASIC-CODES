### PYTHON PROGRAM TO FIND THE FACTORIAL OF A NUMBER 

```python
num = int(input("enter the number "))
fact = 1
if num < 0:
    print("negative number dont have factorial")
elif num == 0 :
    print(f"the factorial of 0 is {1}")
elif num > 0:
    for i in range (1,num+1):
        fact = fact*i
    print(f" the factorial of the number {num} is {fact}")
```

```python
#  FINDING FACTORIAL THROUGH RECURSION

num=int(input("enter the number "))
def fact(a):
    if a == 0:
        return 1
    else:
        return((a*fact(a-1)))
result=fact(num)
print(f"the factorial of the number {num} is {result}")
    
```
