## How to Parse a String to a Float or Integer using Python Codes?

```python
# converstion of string to integer \
    
string = "12345"
print(type(string))
int_str = int(string)
print(type(int_str))
```

```python
# converstion of string into float 

string = "933.02"
print(type(str))
float_str = float(string)
print(type(float_str))
```

```python
# converstion of the string_float into integer 

string = "1234.09"
print(type(string))
str_float_int = int(float(string))
print(type(str_float_int))
```
